```python
import sys
import time
from grammar.contentchecker import check_content, SentenceRequirement, WordType
from feedback.Feedback import get_feedback
import transcription.transcription as transcription
```

```python
test_audio = "benchmarks/audio.wav"
test_incorrect_sentence = "De beer eten honing."
test_correct_sentence = "De beer eet honing."

content_requirements = [
    SentenceRequirement(
        type=WordType.noun,
        options=["beer", "beertje"]
    ),
    SentenceRequirement(
        type=WordType.noun,
        options=["honing"]
    )
]
```

```python
get_feedback(test_incorrect_sentence)
```

```python
with open("benchmarks/audio.wav", "rb") as f:
    audio_file_data = f.read()

def transcribe():
    return transcription.transcribe_audio(audio_file_data)

def check_content_wrapper():
    return check_content(test_correct_sentence.split(), content_requirements)
    
def get_feedback_incorrect():
    return get_feedback(test_incorrect_sentence)

def get_feedback_correct():
    return get_feedback(test_correct_sentence)

def benchmark(function):
    start_time = time.time()
    result = function()
    end_time = time.time()
    duration = end_time - start_time
    return duration

transcription_durations = [] 
check_content_durations = []
get_feedback_correct_durations = []
get_feedback_incorrect_durations = []

for i in range(5):
    transcription_duration = benchmark(transcribe)
    transcription_durations.append(transcription_duration)

    print(f"Transcription duration: {transcription_duration:.2f} seconds")

    check_content_duration = benchmark(check_content_wrapper)
    check_content_durations.append(check_content_duration)

    get_feedback_correct_duration = benchmark(get_feedback_correct)
    get_feedback_correct_durations.append(get_feedback_correct_duration)

    get_feedback_incorrect_duration = benchmark(get_feedback_incorrect)
    get_feedback_incorrect_durations.append(get_feedback_incorrect_duration)

print(f"Average transcription duration: {sum(transcription_durations) / len(transcription_durations):.2f} seconds")
print(f"Average check content duration: {sum(check_content_durations) / len(check_content_durations):.2f} seconds")
print(f"Average get feedback correct duration: {sum(get_feedback_correct_durations) / len(get_feedback_correct_durations):.2f} seconds")
print(f"Average get feedback incorrect duration: {sum(get_feedback_incorrect_durations) / len(get_feedback_incorrect_durations):.2f} seconds")
```
